```python
import pandas as pd
from pycaret.datasets import get_data

# Pobieramy wbudowany zbiór danych 'income' z biblioteki PyCaret
df = get_data('income')

# Wyświetlamy pierwsze 5 wierszy, aby zobaczyć strukturę tabeli
df.head()
```

```python
# 1. Sprawdzamy wymiary tabeli (liczba wierszy, liczba kolumn)
print("Wymiary zbioru (wiersze, kolumny):", df.shape)

# 2. Wyświetlamy typy danych oraz liczbę niepustych wartości w każdej kolumnie
df.info()
```

```python
from pycaret.classification import setup

# Inicjalizacja środowiska eksperymentu
clf_setup = setup(
    data=df, 
    target='income >50K', 
    session_id=123
)
```

```python
from pycaret.classification import *

best_model = compare_models()
```

```python
from pycaret.classification import tune_model

# Optymalizujemy hiperparametry naszego najlepszego modelu
tuned_model = tune_model(best_model)
```

```python
from pycaret.classification import save_model

# Zapisujemy bazowy, najlepszy model wraz z całym pipeline'em przetwarzania danych
save_model(best_model, 'income_pipeline')
```
